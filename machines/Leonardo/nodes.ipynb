```python
import pandas as pd
import re
```

```python
def parse_distance_matrix(path: str) -> pd.DataFrame:
    df = pd.read_csv(
        path,
        skip_blank_lines=True
    )
    return df

df = parse_distance_matrix("distance_matrix.csv")
df
```

```python
def parse_topology_file(path: str) -> pd.DataFrame:
    # Read file while ignoring lines that start with '#'
    df = pd.read_csv(
        path,
        delim_whitespace=True,     # Split by any whitespace
        comment='#',               # Ignore lines starting with '#'
        names=['Nodes', 'IswL1s', 'Cells'],  # Column names (since header line is commented)
        skip_blank_lines=True
    )
    return df

df = parse_topology_file("ntopology.dat")
df
```

```python
def parse_topology_file_v2(path: str) -> pd.DataFrame:
    pattern = re.compile(
        r"NODE\s+(\d+)\s+RACK\s+(\d+)\s+CELL\s+(\d+)\s+ROW\s+(\d+)\s+PARTITION\s+(\d+)\s+SWITCH\s+(\d+)"
    )
    
    records = []
    with open(path) as f:
        for line in f:
            match = pattern.search(line)
            if match:
                records.append(match.groups())
    
    df = pd.DataFrame(
        records,
        columns=["NODE", "RACK", "CELL", "ROW", "PARTITION", "SWITCH"]
    ).astype(int)
    
    return df

df = parse_topology_file_v2("leo_map.txt")

for p in df['PARTITION'].unique():
    partition_df = df[df['PARTITION'] == p]
    print(f'=========== Partition: {p} ===========')
    for c in partition_df.columns:
        print(f'{c} (len {len(partition_df[c].unique())}): {partition_df[c].unique()}')
```
